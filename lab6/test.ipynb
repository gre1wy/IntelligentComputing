```python
from transformers import pipeline
```

```python

generator = pipeline("text-generation", model="flax-community/t5-recipe-generation")
```

```python
ingredients = "chicken, salt, pepper"
prompt = f"Generate a recipe using {ingredients}:"

recipe = generator(prompt, max_length=500)
print(recipe[0]["generated_text"])
```
